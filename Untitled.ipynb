```python
import pkg_resources

# ambil semua paket terinstall
installed_packages = pkg_resources.working_set

# urutkan berdasarkan nama paket
packages = sorted([f"{pkg.key}=={pkg.version}" for pkg in installed_packages])

# tulis ke file requirements.txt
with open("requirements.txt", "w") as f:
    for pkg in packages:
        f.write(pkg + "\n")

print(f"Berhasil generate requirements.txt dengan {len(packages)} paket.")
```
